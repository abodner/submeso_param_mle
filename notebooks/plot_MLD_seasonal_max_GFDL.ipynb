```python
# load netcdf from OM4 runs

import numpy as np
import matplotlib.pyplot as plt
import os
import xarray as xr
import cartopy.crs as ccrs
from matplotlib.colorbar import ColorbarBase
```

```python
# paths to dataset
BASE = '/orcd/data/abodner/002/shared_datasets/om4_output/new_runs/'


# load FK MLD data 
ds_MLD_EN1_FK = xr.open_dataset(BASE+'BFF11/ocean_monthly.201801-202212.MLD_EN1.nc')
ds_MLD_EN2_FK = xr.open_dataset(BASE+'BFF11/ocean_monthly.201801-202212.MLD_EN2.nc')

# group by season and take mean over summer (EN1) and winter (EN2) for last 5 years of JRA cycle
# Southern Hemisphere (SH)
MLD_summer_SH_FK = ds_MLD_EN1_FK.MLD_EN1.groupby("time.season").max('time',skipna=True).sel(season='DJF').isel(yh=range(0,503))
MLD_winter_SH_FK = ds_MLD_EN2_FK.MLD_EN2.groupby("time.season").max('time',skipna=True).sel(season='JJA').isel(yh=range(0,503))

# Northern Hemisphere (NH)
MLD_summer_NH_FK = ds_MLD_EN1_FK.MLD_EN1.groupby("time.season").max('time',skipna=True).sel(season='JJA').isel(yh=range(503,1080))
MLD_winter_NH_FK = ds_MLD_EN2_FK.MLD_EN2.groupby("time.season").max('time',skipna=True).sel(season='DJF').isel(yh=range(503,1080))

# concat SH and NH for summer and winter datasets
MLD_summer_FK = xr.concat([MLD_summer_SH_FK, MLD_summer_NH_FK], "yh")
MLD_winter_FK = xr.concat([MLD_winter_SH_FK, MLD_winter_NH_FK], "yh")


# load BD MLD data 
ds_MLD_EN1_BD = xr.open_dataset(BASE+'BOD23/ocean_monthly.201801-202212.MLD_EN1.nc')
ds_MLD_EN2_BD = xr.open_dataset(BASE+'BOD23/ocean_monthly.201801-202212.MLD_EN2.nc')

# group by season and take mean over summer (EN1) and winter (EN2) for last 5 years of JRA cycle
# Southern Hemisphere (SH)
MLD_summer_SH_BD = ds_MLD_EN1_BD.MLD_EN1.groupby("time.season").max('time',skipna=True).sel(season='DJF').isel(yh=range(0,503))
MLD_winter_SH_BD = ds_MLD_EN2_BD.MLD_EN2.groupby("time.season").max('time',skipna=True).sel(season='JJA').isel(yh=range(0,503))

# Northern Hemisphere (NH)
MLD_summer_NH_BD = ds_MLD_EN1_BD.MLD_EN1.groupby("time.season").max('time',skipna=True).sel(season='JJA').isel(yh=range(503,1080))
MLD_winter_NH_BD = ds_MLD_EN2_BD.MLD_EN2.groupby("time.season").max('time',skipna=True).sel(season='DJF').isel(yh=range(503,1080))

# concat SH and NH for summer and winter datasets
MLD_summer_BD = xr.concat([MLD_summer_SH_BD, MLD_summer_NH_BD], "yh")
MLD_winter_BD = xr.concat([MLD_winter_SH_BD, MLD_winter_NH_BD], "yh")





# load no-MLE MLD data 
ds_MLD_EN1_NO = xr.open_dataset(BASE+'noMLE/ocean_monthly.201801-202212.MLD_EN1.nc')
ds_MLD_EN2_NO = xr.open_dataset(BASE+'noMLE/ocean_monthly.201801-202212.MLD_EN2.nc')

# group by season and take mean over summer (EN1) and winter (EN2) for last 5 years of JRA cycle
# Southern Hemisphere (SH)
MLD_summer_SH_NO = ds_MLD_EN1_NO.MLD_EN1.groupby("time.season").max('time',skipna=True).sel(season='DJF').isel(yh=range(0,503))
MLD_winter_SH_NO = ds_MLD_EN2_NO.MLD_EN2.groupby("time.season").max('time',skipna=True).sel(season='JJA').isel(yh=range(0,503))

# Northern Hemisphere (NH)
MLD_summer_NH_NO = ds_MLD_EN1_NO.MLD_EN1.groupby("time.season").max('time',skipna=True).sel(season='JJA').isel(yh=range(503,1080))
MLD_winter_NH_NO = ds_MLD_EN2_NO.MLD_EN2.groupby("time.season").max('time',skipna=True).sel(season='DJF').isel(yh=range(503,1080))

# concat SH and NH for summer and winter datasets
GFDL_MLD_noMLE_summer = xr.concat([MLD_summer_SH_NO, MLD_summer_NH_NO], "yh")
GFDL_MLD_noMLE_winter = xr.concat([MLD_winter_SH_NO, MLD_winter_NH_NO], "yh")



# load static file
#ds_stat = xr.open_dataset(BASE+'ocean_annual_z.static.nc')
```

```python
# paths to dataset
BASE = '/orcd/data/abodner/002/shared_datasets/om4_output/'


# load ARGO MLD data 
ds_MLD_EN1_ARGO = xr.open_dataset(BASE+'Argo_MLD.mld_pe_anomaly_25.nc')
ds_MLD_EN2_ARGO = xr.open_dataset(BASE+'Argo_MLD.mld_pe_anomaly_2500.nc')

# group by season and take mean over summer (EN1) and winter (EN2) for last 5 years of JRA cycle
# Southern Hemisphere (SH)
MLD_summer_SH_ARGO = ds_MLD_EN1_ARGO.MLD_mean.sel(Month=[12,1,2]).max('Month',skipna=True).sel(Lat=slice(-90,0))
MLD_winter_SH_ARGO = ds_MLD_EN2_ARGO.MLD_mean.sel(Month=[6,7,8]).max('Month',skipna=True).sel(Lat=slice(-90,0))

                                                                           
# Northern Hemisphere (NH)
MLD_summer_NH_ARGO = ds_MLD_EN1_ARGO.MLD_mean.sel(Month=[6,7,8]).max('Month',skipna=True).sel(Lat=slice(0,90))
MLD_winter_NH_ARGO = ds_MLD_EN2_ARGO.MLD_mean.sel(Month=[12,1,2]).max('Month',skipna=True).sel(Lat=slice(0,90))

# concat SH and NH for summer and winter datasets
MLD_summer_ARGO = xr.concat([MLD_summer_SH_ARGO, MLD_summer_NH_ARGO], "Lat")
MLD_winter_ARGO = xr.concat([MLD_winter_SH_ARGO, MLD_winter_NH_ARGO], "Lat")


# interpolate ARGO data to the same grid as OM4 (first need to shift lon to -300:60

MLD_summer_ARGO_interp = MLD_summer_ARGO.roll(Lon=150).assign_coords(Lon=(MLD_summer_ARGO.Lon - 300)).interp(Lon=MLD_winter_BD.xh, Lat = MLD_winter_BD.yh, method="linear")
MLD_winter_ARGO_interp = MLD_winter_ARGO.roll(Lon=150).assign_coords(Lon=(MLD_winter_ARGO.Lon - 300)).interp(Lon=MLD_winter_BD.xh, Lat = MLD_winter_BD.yh, method="linear")
```
